### 📌 Project Overview
**NOVORA** is a Machine Learning-driven predictive analytics framework designed to estimate student academic outcomes using historical educational and behavioral attributes. The framework leverages data-driven methodologies to identify key factors influencing academic achievement, empowering educators and institutional leaders to deliver targeted interventions.

## 🎯 Phase 1: Problem Understanding

### 1.1 Objective
To formulate and architect a supervised machine learning solution capable of estimating student academic performance (`Exam_Score`) based on measurable behavioral and instructional indicators.

### 1.2 Problem Statement
Educational institutions amass vast volumes of administrative and academic metrics. However, identifying at-risk students who require timely academic support or guidance before final summative evaluations remains a major operational challenge.

By deploying Machine Learning algorithms, historical student engagement indicators can uncover latent patterns of student success and deliver reliable early warnings.

### 1.3 Key Predictive Indicators
- **Weekly Self Study Hours (`Study_Hours`)**: Dedicated independent study outside lecture hours.
- **Attendance Percentage (`Attendance`)**: Class participation and presence over the instructional term.
- **Coursework Completion (`Assignments_Completed`)**: Continuous formative assessment submissions.
- **Target Outcome (`Exam_Score`)**: Comprehensive final academic performance metric (0-100 scale).

### 1.4 Expected Outcomes
1. **Accurate Performance Forecasting**: High explanatory power ($R^2$) with minimized prediction error (MAE/RMSE).
2. **Data-Driven Intervention Support**: Assisting educators in identifying students needing mentorship before failure occurs.
3. **Evidence-Based Academic Governance**: Transparent educational monitoring and policy refinement.

## 📦 Phase 2: Data Collection & Profiling

### 2.1 Objective
To collect, import, verify, and statistically characterize the foundational dataset for the machine learning pipeline.

### 2.2 Operational Activities Performed
1. Ingest dataset into the analytical environment.
2. Verify file integrity and successful loading.
3. Audit dataset dimensions, matrix shape, and memory consumption.
4. Classify attribute types and schema definitions.
5. Compute descriptive statistical parameters (central tendency, dispersion, shape).
6. Conduct quality checks: null values, duplicates, and outlier identification.
7. Examine feature inter-correlations.

```python
# Import core analytical and visualization libraries
import os
import hashlib
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Set visual style
sns.set_theme(style="whitegrid", palette="muted")
%matplotlib inline

print("Environment initialized successfully.")
```

### 2.3 Step 1: Dataset Import & Integrity Verification

```python
# Dataset Path definition
dataset_path = os.path.join("..", "data", "student_performance_data.csv")

# File verification
assert os.path.exists(dataset_path), f"File not found at {dataset_path}"
file_size = os.path.getsize(dataset_path)

with open(dataset_path, 'rb') as f:
    checksum = hashlib.md5(f.read()).hexdigest()

print(f"File Path:     {dataset_path}")
print(f"File Size:     {file_size} bytes ({file_size/1024:.2f} KB)")
print(f"MD5 Checksum:  {checksum}")

# Ingest into Pandas DataFrame
df = pd.read_csv(dataset_path)
print("\n[+] Dataset successfully loaded into DataFrame.")
df.head(10)
```

### 2.4 Step 2: Dataset Dimensions & Structural Schema Audit

```python
rows, cols = df.shape
print(f"Total Number of Rows (Student Instances): {rows}")
print(f"Total Number of Columns (Attributes):     {cols}")
print(f"Total Data Points:                        {rows * cols}")
print(f"DataFrame Memory Footprint:               {df.memory_usage(deep=True).sum() / 1024:.2f} KB")
print(f"Duplicate Rows:                           {df.duplicated().sum()}")

print("\nDataFrame Info:")
df.info()
```

### 2.5 Step 3: Data Quality & Missing Value Inspection

```python
quality_df = pd.DataFrame({
    "Data Type": df.dtypes,
    "Non-Null Count": df.notnull().sum(),
    "Null Count": df.isnull().sum(),
    "Null Percentage (%)": (df.isnull().sum() / len(df)) * 100,
    "Unique Values": df.nunique()
})
quality_df
```

### 2.6 Step 4: Descriptive Statistical Summaries

```python
# Comprehensive Statistical Profiling
stats_summary = pd.DataFrame({
    "Mean": df.mean(),
    "Std Dev": df.std(),
    "Median": df.median(),
    "Min": df.min(),
    "Q25": df.quantile(0.25),
    "Q75": df.quantile(0.75),
    "IQR": df.quantile(0.75) - df.quantile(0.25),
    "Max": df.max(),
    "Skewness": df.skew(),
    "Kurtosis": df.kurtosis()
})
stats_summary.round(3)
```

### 2.7 Step 5: Feature Correlation Analysis

```python
corr = df.corr()
print("Correlation with Target Feature (Exam_Score):")
print(corr["Exam_Score"].sort_values(ascending=False))

plt.figure(figsize=(7, 5))
sns.heatmap(corr, annot=True, cmap="coolwarm", fmt=".3f", vmin=-1, vmax=1, linewidths=1)
plt.title("NOVORA: Feature Correlation Matrix", fontsize=13, fontweight='bold')
plt.tight_layout()
plt.show()
```

### 2.8 Step 6: Distribution and Outlier Visualizations

```python
fig, axes = plt.subplots(2, 2, figsize=(12, 8))
axes = axes.flatten()
colors = ['#1f77b4', '#2ca02c', '#ff7f0e', '#d62728']

for i, col in enumerate(df.columns):
    sns.histplot(df[col], kde=True, ax=axes[i], color=colors[i], edgecolor='black', alpha=0.6)
    axes[i].set_title(f"Distribution: {col}", fontweight='bold')
    axes[i].axvline(df[col].mean(), color='red', linestyle='--', label=f"Mean: {df[col].mean():.2f}")
    axes[i].axvline(df[col].median(), color='green', linestyle=':', label=f"Median: {df[col].median():.2f}")
    axes[i].legend()

plt.suptitle("Feature Distributions with KDE", fontsize=14, fontweight='bold')
plt.tight_layout()
plt.show()
```

### 2.9 Step 7: Bivariate Relationship with Target (`Exam_Score`)

```python
fig, axes = plt.subplots(1, 3, figsize=(15, 4.5))
predictors = ['Study_Hours', 'Attendance', 'Assignments_Completed']

for i, col in enumerate(predictors):
    sns.regplot(data=df, x=col, y='Exam_Score', ax=axes[i], scatter_kws={'alpha': 0.6, 'color': colors[i]}, line_kws={'color': 'darkred'})
    axes[i].set_title(f"{col} vs Exam_Score", fontweight='bold')

plt.suptitle("Regression Scatterplots with Exam_Score", fontsize=14, fontweight='bold')
plt.tight_layout()
plt.show()
```
